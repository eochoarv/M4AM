```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = 1 - 2*np.random.rand(3,2)
plt.scatter(x[:,0], x[:,1])
```

```python
theta = 1 - 2*np.random.rand(100,2)
```

```python
A = theta @ x.T
```

```python
np.quantile(A[:,0], .8)
```

```python
plt.scatter(theta[:,0], theta[:,1], c=A[:,0]>.3, alpha=.5)
plt.scatter(theta[:,0], theta[:,1], c=A[:,1]>.3, alpha=.5)
plt.scatter(theta[:,0], theta[:,1], c=A[:,2]>.3, alpha=.5)
```

```python
plt.scatter(theta[:,0], theta[:,1], c=A[:,0]>.3)
plt.scatter(x[:,0], x[:,1], c='r')
```

```python
plt.scatter(theta[:,0], theta[:,1], c=A[:,1]>.3)
plt.scatter(x[:,0], x[:,1], c='r')
```

```python
plt.scatter(theta[:,0], theta[:,1], c=A[:,2]>.3)
plt.scatter(x[:,0], x[:,1], c='r')
```
