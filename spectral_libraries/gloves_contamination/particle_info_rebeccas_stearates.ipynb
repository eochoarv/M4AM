```python
import os
import numpy as np
import pandas as pd
from sklearn.metrics import pairwise_distances
```

```python
path = '../Reference_library_reb/'
dirs = os.listdir(path)
dirs = [x for x in dirs if x not in ['.DS_Store', 'Archive for reference data', 'Processed']]

file_list = []
for d in dirs:
    path_2 = path + d + '/Data/Raw/mIRage OPTIR+Raman/'
    dirs_2 = os.listdir(path_2)
    file_list_aux = [x for x in dirs_2 if x[-3:]=="csv"]
    dirs_2 = [[x+'/'+y for y in os.listdir(path_2 + x)] for x in dirs_2 if os.path.isdir(path_2 + x)]
    dirs_3 = []
    if len(dirs_2) > 0:
        for d2 in dirs_2:
            dirs_3 = dirs_3 + d2
    file_list_aux = file_list_aux + [x for x in dirs_3 if x[-3:]=="csv"]
    file_list_aux = [path_2 + x for x in file_list_aux]
    file_list = file_list + file_list_aux 
```

```python
label = ['PP' if 'PP' in x else 
         'PE' if 'PE' in x else 
         'PSL' if 'PSL' in x else 
         'sucrose' if 'sucrose' in x else 
         'NH2SO4' if 'NH2SO4' in x else 'x' for x in file_list]

```

```python
df_list = []

for f, l in zip(file_list, label):
    df = pd.read_csv(f)
    if df.shape[1] in [5, 7]:
        df = df.iloc[:,:5]
        df.columns = ['Feature ID', 'X', 'Y', 'Size', 'Area']
        df['label'] = l
        df_list.append(df.reset_index(drop=True))
    
pd.concat(df_list,axis=0)
```

```python
path = '../MEC_04_010/Raw Data/Stearates/'
dirs = os.listdir(path)

for d in dirs:
    if d[-4:] == 'ptir' or d==".DS_Store":
        del dirs[np.where(np.array(dirs) == d)[0][0]]
        
files_list = []
for d in dirs:
    files = os.listdir(path + d)
    files_list = files_list + [d+'/'+f for f in files if f[-3:] == 'csv']
```

```python
df_list = []
for f in files_list:
    df = pd.read_csv(path + f)
    df = df.rename({'X (um)':'map_x', 'Y (um)':'map_y'}, axis=1)
    df['polymer'] = f.split('/')[0]
    df_list.append(df)
    
df_all = pd.concat(df_list, axis=0).reset_index(drop=True)
df_all
```

```python
preposs_data_ir = pd.read_csv('stearates_data_ir_final.csv').iloc[:,-4:]
preposs_data_ir
```

```python
spectra_matches_list = []
for p in np.unique(preposs_data_ir['polymer']):
    
    #if coord_df.query('id_file == @id_').shape[0] == 0:
    #    continue
    id_matches = np.argmin(pairwise_distances(preposs_data_ir.query('polymer == @p').loc[:,['map_x', 'map_y']], 
                                              df_all.query('polymer == @p').loc[:,['map_x', 'map_y']]),1)


    coord_matches = df_all.query('polymer == @p').iloc[id_matches, :].reset_index(drop=True)
    spectra_matches = (preposs_data_ir.
                       query('polymer == @p').
                       reset_index(drop=True).
                       merge(coord_matches, left_index=True, right_index=True))
    
    spectra_matches['distance'] = np.sqrt((spectra_matches['map_x_x'] - spectra_matches['map_x_y']) ** 2 + 
                                          (spectra_matches['map_y_x'] - spectra_matches['map_y_y']) ** 2)
    spectra_matches_list.append(spectra_matches)
    #coord_matches = coord_df.iloc[id_matches, :].reset_index(drop=True)
    
spectra_matches_all = pd.concat(spectra_matches_list).reset_index(drop=True)
spectra_matches_all.loc[:, ['id', 'Size (um)', 'Area (um²)']].to_csv('particle_info_stearate.csv')
```
