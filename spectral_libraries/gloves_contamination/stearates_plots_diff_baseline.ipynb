```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
preposs_data_raman_als = pd.read_csv('stearates_preprocessed_CS_raman_als.csv')
del preposs_data_raman_als['Unnamed: 0']
preposs_data_raman_als

preposs_data_raman_rubber = pd.read_csv('stearates_preprocessed_CS_raman_rubber.csv')
del preposs_data_raman_rubber['Unnamed: 0']
preposs_data_raman_rubber

preposs_data_raman_specy = pd.read_csv('stearates_preprocessed_CS_raman_specy.csv')
del preposs_data_raman_specy['Unnamed: 0']
preposs_data_raman_specy

preposs_data_raman_none = pd.read_csv('stearates_preprocessed_CS_raman_none.csv')
del preposs_data_raman_none['Unnamed: 0']
preposs_data_raman_none
```

```python
wn_raman = np.float64(preposs_data_raman_als.columns[:-4])

idx_raman = np.logical_and(wn_raman > 355, wn_raman < 3800)

X_raman_als = np.array(preposs_data_raman_als.iloc[:,:-4])
X_raman_rubber = np.array(preposs_data_raman_rubber.iloc[:,:-4])
X_raman_specy = np.array(preposs_data_raman_specy.iloc[:,:-4])
X_raman_specy = (X_raman_specy - X_raman_specy.mean(1)[:,None]) / X_raman_specy.std(1)[:,None]
X_raman_none = np.array(preposs_data_raman_none.iloc[:,:-4])
```

# No baseline correction

```python
_ = plt.plot(wn_raman[idx_raman],
             X_raman_none[:,idx_raman].T)
```

# ALS baseline

```python
_ = plt.plot(wn_raman[idx_raman],
             X_raman_als[:,idx_raman].T)
```

# Rubberband baseline

```python
_ = plt.plot(wn_raman[idx_raman],
             X_raman_rubber[:,idx_raman].T)
```

# Specy baseline

```python
_ = plt.plot(wn_raman[idx_raman],
             X_raman_specy[:,idx_raman].T)
```

```python
n_peaks = 7
diff_wn = 30

for j in range(X_raman_none.shape[0]):
    idx_sort = np.argsort(X_raman_none[:,:][j,:])[::-1]
    idx_decre = np.where(np.diff(X_raman_none[j,:][:-1])<0)[0]
    idx_incre = np.where(np.diff(X_raman_none[j,:][:-1])>0)[0]+1
    peaks = np.array([i for i in idx_sort if i in idx_decre and i in idx_incre])
    
    _id = preposs_data_raman_none['id'][j]
    print(_id)
    #print('coords: ' + str((coord_x, coord_y)))
    
    _ = plt.plot(wn_raman,
                 X_raman_none[j,:].T, label='no baseline correction')
    
    _ = plt.plot(wn_raman,
                 X_raman_als[j,:].T-7, alpha=.5, label='als baseline correction')
    
    _ = plt.plot(wn_raman,
                 X_raman_rubber[j,:].T-14, alpha=.5, label='rubber baseline correction')
    
    _ = plt.plot(wn_raman,
                 X_raman_specy[j,:].T-21, alpha=.5, label='specy baseline correction')
    
    for n, i in enumerate(peaks[:n_peaks]):
        if n > 0 and np.min(np.abs(wn_raman[peaks[:n]] - wn_raman[peaks[n]])) < diff_wn:
            #print(wn[idx][peaks[n]])
            continue
        plt.scatter(wn_raman[i], 
                    X_raman_none[j,:][i], c='r')
        
        plt.annotate(text=str(int(wn_raman[i])), 
                     xy=(wn_raman[i], X_raman_none[j,:][i]),
                     xytext=(wn_raman[i], X_raman_none[j,:][i]+1))
        
    _ = plt.legend(loc=2)
    _ = plt.ylim(-25, 15)
    _ = plt.show()
```
