```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path1 = '/Users/eochoa/Downloads/usgs_splib07/ASCIIdata/ASCIIdata_splib07a/'
path = '/Users/eochoa/Downloads/usgs_splib07/ASCIIdata/ASCIIdata_splib07a/ChapterO_OrganicCompounds/'
files = os.listdir(path)
files
```

```python
table = np.cumsum(pd.read_csv(path1 + 'splib07a_Wavelengths_ASD_0.35-2.5_microns_2151_ch.txt'))
table.columns = ['wavelength']

for f in files:
    t = pd.read_csv(path + f, skiprows=0)
    t.columns = [f]
    table = table.merge(t, left_index=True, right_index=True)

table = table.set_index('wavelength')
table
```
