```python
import spectrochempy as scp
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
```

```python
dir_paht = 'Paint_MP_library/spa files/'
files_list = os.listdir(dir_paht)
```

```python
X = [scp.read_omnic(dir_paht+x) for x in os.listdir(dir_paht)]
```

```python
inventory = pd.read_excel('Paint_MP_library/paint-library-inventory.xlsx', skiprows=[0])
inventory = inventory.loc[:,['Polymer','.SPA file name']]
inventory['.SPA file name'] = inventory['.SPA file name'].str.lower()
```

```python
inventory.groupby('Polymer').count()
```

```python
df = pd.DataFrame({X[i].name+'.spa':X[i].data[0] for i in range(21)})
df['wavelength'] = X[0].x.data
df.columns = df.columns.str.lower()
df = df.melt('wavelength')
df = pd.merge(df, inventory, how='left', left_on='variable', right_on='.SPA file name')
df = df.drop(columns='.SPA file name')
df.columns = ['wavelength', 'file_name', 'intensity', 'polymer']
```

```python
df.groupby('polymer').count()
```
