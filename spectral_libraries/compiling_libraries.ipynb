```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
local_path = '/Users/eochoa/Documents/Research_Abuj/M4AM/spectral_libraries/'
samples_ma_dict = pd.read_csv('./Mendeley_data/samples_ multil_assembly_dict.csv')
polymer_dict = pd.read_csv('./Mendeley_data/polymer_dict.csv')

path = './Mendeley_data/83my5tpmpw-2/'

file_name_list = os.listdir(path=path)

df_list = list()
for f in file_name_list:
    
    df = pd.read_csv(path + f, sep = ',', header=None, skiprows=5)
    df.insert(2, "file_name", [f]*len(df), True) 
    df = df.loc[:, [0, 1, 'file_name']]
    df_list.append(df)
    
df_spectra = pd.concat(df_list, axis=0)
df_spectra.columns = ['wavelength', 'intensity', 'file_name']

samples_ma_dict = pd.melt(samples_ma_dict, id_vars='sample')
samples_ma_dict.rename(columns={'variable':'multil_assembly', 
                                'value':'polymer_short'}, inplace=True)
samples_ma_dict = pd.merge(samples_ma_dict, polymer_dict, how='left')

file_name_split = [x.split() for x in np.sort(file_name_list)]
file_name_df = pd.DataFrame([x[0:2] if (x[0][0:2] != '20' and x[1][:3] != 'mid') 
 else [x[0], 'ml' + x[2][0]] if (x[0][0:2] != '20' and x[1] == 'midlayer')
 else [x[0], 'ml' + x[3]] if (x[0][0:2] != '20' and x[1] == 'mid')
 else [x[1][1:], x[2][0:x[2].find('_')]] if x[2][0:x[2].find('_')] != 'printed.1'
 else [x[1][1:], x[2][0:x[2].find('.')]] for x in file_name_split])

file_name_df.columns = ['sample', 'multil_assembly']
file_name_df['file_name'] = np.sort(file_name_list)
file_name_df = file_name_df.astype({'sample':int})
file_name_df = pd.merge(file_name_df, samples_ma_dict, on=['sample', 'multil_assembly'])

df_spectra_1 = pd.merge(df_spectra, file_name_df.loc[:,['file_name', 'polymer']])
df_spectra_1['data_source'] = 'food_packaging'
df_spectra_1['type_sprectra'] = 'FTIR'
```
