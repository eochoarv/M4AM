```python
X = [scp.read_omnic(dir_paht + x) for x in files_list]
```

```python
[X[i].name+'-'+poly[i] for i in range(len(X))]
```

```python
df = pd.DataFrame({X[i].name+'.'+str(i)+'_'+poly[i]:X[i].data[0] for i in range(len(X))})
df['wavelength'] = X[0].x.data
df.columns = df.columns.str.lower()
df = df.melt('wavelength')
df['polymer'] = [x[(x.find('_')+1):] for x in df['variable']]
df.columns = ['wavelength', 'file_name', 'intensity', 'polymer']
```
