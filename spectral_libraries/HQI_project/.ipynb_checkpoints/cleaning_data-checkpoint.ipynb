```python
X = [scp.read_omnic(dir_paht + x) for x in files_list]
```

```python
df = pd.DataFrame({X[i].name+'.spa':X[i].data[0] for i in range(5)})
df['wavelength'] = X[0].x.data
df.columns = df.columns.str.lower()
df = df.melt('wavelength')
df['polymer'] = 'sea_sand'
#df = pd.merge(df, inventory, how='left', left_on='variable', right_on='.SPA file name')
#df = df.drop(columns='.SPA file name')
df.columns = ['wavelength', 'file_name', 'intensity', 'polymer']
df.to_csv('sea_sand.csv')
```
