```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
not_truncated = pd.read_csv('results_final/20251209/not_truncated_id_mix_53_ir.csv')
del not_truncated['Unnamed: 0']

truncated = pd.read_csv('results_final/20251209/truncated_id_mix_53_ir.csv')
del truncated['Unnamed: 0']


truncated.columns = list(np.float64(truncated.columns))
not_truncated.columns = list(np.float64(not_truncated.columns))
```

```python
x = truncated
x.columns = np.float64(x.columns)
                
x = x.T.reset_index()
x.columns = ['wn', 'signal']
wn = np.arange(947.0, 2999.0, 2)
x = pd.DataFrame({'wn':wn}).merge(x, on='wn', how='left').set_index('wn').T
                

y = not_truncated
y.columns = np.float64(y.columns)
                
y = y.T.reset_index()
y.columns = ['wn', 'signal']
wn = np.arange(947.0, 2999.0, 2)
y = pd.DataFrame({'wn':wn}).merge(y, on='wn', how='left').set_index('wn').T
```

```python
x = x.reset_index(drop=True)
y = y.reset_index(drop=True)
```

```python
spectra = pd.concat([y, x], axis=0)
spectra.to_csv('results_final/20251209/id_mix_53.csv')
```
