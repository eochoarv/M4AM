```python
import os
import numpy as np
import pandas as pd
from collections import defaultdict
from tqdm import trange, tqdm

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.cluster import KMeans

from utils import *
```

```python
spectra_data = pd.read_csv('OPTIR/Pilot_Phase_Field_Samples/data_all.csv')
del spectra_data['Unnamed: 0']

#spectra_data_long = spectra_data.iloc[:, :-2].reset_index().melt(id_vars='index').astype('float32')
spectra_data_long = spectra_data.iloc[:, :-1].melt(id_vars='id')#
spectra_data_long.iloc[:,1:] = spectra_data_long.iloc[:,1:].astype('float32')

spectra_data_long.columns = ['file_name','wavelength','intensity']
spectra_data_long['polymer'] = ''
spectra_data_long_int = interpol_spect(spectra_data_long, min_wl=674.9918, max_wl=3999.809,n=1725)
spectra_data_int = spectra_data_long_int.pivot(
    index=['file_name'], 
    columns='wavelength', 
    values='intensity').reset_index()

spectra_data_np = np.float32(spectra_data_int.iloc[:,1:].to_numpy())
spectra_data_np_norm = (spectra_data_np[:,idx_range] - spectra_data_np[:,idx_range].mean(1)[:,None]) / spectra_data_np[:,idx_range].std(1)[:,None]
```

```python
spectra_data_std = pd.read_csv('OPTIR/Rebecca_standards/IR_standards.csv')
spectra_data_std = spectra_data_std.melt(id_vars='wavelength')
spectra_data_std['polymer'] = ''
spectra_data_std.columns = ['wavelength', 'file_name', 'intensity', 'polymer']
spectra_data_std = norm_spect(spectra_data_std, norm_type='SNV')
spectra_data_std_int = interpol_spect(spectra_data_std, min_wl=674.9918, max_wl=3999.809,n=1725)
spectra_data_std_int = spectra_data_std_int.pivot(
    index=['file_name'], 
    columns='wavelength', 
    values='intensity').reset_index()
#spectra_data_std_int = (spectra_data_std_int.iloc[:,1:] - spectra_data_std_int.iloc[:,1:].mean()) / (1*(spectra_data_std_int.iloc[:,1:].std() == 0) + spectra_data_std_int.iloc[:,1:].std())
x_std = np.float32(spectra_data_std_int.iloc[:,1:].to_numpy())
x_std
```

```python
df_library = pd.read_csv('spectral_data_all.csv')
df_library.drop(columns=['Unnamed: 0'], inplace=True)
df_flopp = df_library.query('data_source == "flopp" or data_source == "flopp_e"').reset_index(drop=True)
df_flopp = df_flopp.query('intensity > 0')
df_flopp = norm_spect(df_flopp, norm_type='SNV')
df_flopp_pivot = df_flopp.pivot(index=['file_name', 'polymer', 'data_source'], 
               columns='wavelength', 
               values='intensity').reset_index()
```

```python
wl_list = df_flopp_pivot.columns[3:]
#df_flopp_pivot.columns = list(df_flopp_pivot.columns[:3]) + ['wl_' + str(round(x*100)) for x in df_flopp_pivot.columns[3:]]
polymer_list = df_flopp_pivot['polymer'].drop_duplicates().to_numpy()


#X_train = df_flopp_pivot.query('data_source == "flopp"').iloc[:,3:].reset_index(drop=True)
X_train = df_flopp_pivot.iloc[:,3:].reset_index(drop=True)
X_test = df_flopp_pivot.query('data_source == "flopp_e"').iloc[:,3:].reset_index(drop=True)

#y_train = df_flopp_pivot.query('data_source == "flopp"').iloc[:,1].reset_index(drop=True)
y_train = df_flopp_pivot.iloc[:,1].reset_index(drop=True)
y_test = df_flopp_pivot.query('data_source == "flopp_e"').iloc[:,1].reset_index(drop=True)

idx_range = np.logical_or(
    np.logical_and(
        wl_list > np.float32((spectra_data.columns[:-2])).min(),
        wl_list < np.float32((spectra_data.columns[:-2].to_numpy()[np.where(np.isnan(np.float32(spectra_data.iloc[0,:-2])))])).min()-1, #-1
    ),
    np.logical_and(
        wl_list > np.float32((spectra_data.columns[:-2].to_numpy()[np.where(np.isnan(np.float32(spectra_data.iloc[0,:-2])))])).max()+1,
        wl_list < np.float32((spectra_data.columns[:-2])).max(),
    )
)


#X_train = X_train.reset_index(drop=True)
#y_train = y_train.reset_index(drop=True)
    
x_train = X_train.to_numpy()
y_train_ = y_train.to_numpy()
y_train_[y_train_ == 'Nylon'] = 'Polyamide'
y_train_[y_train_ == 'Polyester'] = 'Polyethylene Terephthalate'

x_train = x_train[:,idx_range]
x_test = X_test.iloc[:,idx_range]

x_train = (x_train - x_train.mean(1)[:, None]) / x_train.std(1)[:, None]
x_test = (x_test - x_test.mean(1)[:, None]) / x_test.std(1)[:, None]

n, p = x_train.shape
```

```python
x_train_diff = np.diff(-x_train)
x_test_diff = np.diff(-x_test)

x_train_diff = (x_train_diff - x_train_diff.mean(1)[:,None]) / x_train_diff.std(1)[:,None]
x_test_diff = (x_test_diff - x_test_diff.mean(1)[:,None]) / x_test_diff.std(1)[:,None]

spectra_data_diff = np.diff(spectra_data_np_norm)
spectra_data_std_diff = np.diff(x_std[:, idx_range])

spectra_data_diff = (spectra_data_diff - spectra_data_diff.mean(1)[:,None]) / spectra_data_diff.std(1)[:,None]
spectra_data_std_diff = (spectra_data_std_diff - spectra_data_std_diff.mean(1)[:,None]) / spectra_data_std_diff.std(1)[:,None]
```

```python
clf = LogisticRegression(random_state=0, C=.1)
clf.fit(x_train_diff, y_train_)
clf.score(x_test_diff, y_test)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import pairwise_distances
```

```python
kmeans = KMeans(n_clusters=19, random_state=0).fit(x_train_diff)
```

```python
kk_centers = kmeans.cluster_centers_
```

```python
pairwise_distances(kk_centers, spectra_data_std_diff).min(0)
```

```python
distance_center_k = pairwise_distances(kk_centers, spectra_data_diff).min(0)
```
