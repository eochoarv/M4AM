```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
#from tqdm import tqdm
import json
from datetime import date
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
```

# IR

```python
directory_name = 'results/20251026/'
file_name_ids = 'Active_PTIR_Only_Rest_Multiple.csv'
file_name_cp_results = 'CP_results_active_unpaired_ir_LikelyMPs.csv'
file_name_spectra_data = 'active_data_ir.csv'
file_name_reference = 'flopp_data_ir_final.csv'
```

```python
df_ids_plot = pd.read_csv(directory_name+file_name_ids)#Passive_PTIR_Only_Rest_Multiple.csv
ids_plot = df_ids_plot['id'].to_numpy()
n = ids_plot.shape[0]

df_match = pd.read_csv(directory_name+file_name_cp_results)
del df_match['Unnamed: 0']
df_match = df_match.query('id in @ids_plot').reset_index(drop=True)

df_spectra = pd.read_csv(directory_name+file_name_spectra_data)
del df_spectra['Unnamed: 0']

df_label_table = pd.read_csv(directory_name+'label_table_ir.csv')
del df_label_table['Unnamed: 0']

df_spectra_ref = pd.read_csv(directory_name+file_name_reference)
del df_spectra_ref['Unnamed: 0']

wn = np.float64(df_spectra_ref.columns[:-4])
```

```python
init_top = 1 # should be at least 1 and less that last_top
last_top = 14 # should be at most 14

with PdfPages(file_name_ids[:-4]+".pdf") as pdf:
    for i in range(n):
        id_ = df_match['id'][i]
        for j in range(init_top-1, last_top):
            poly = df_match.iloc[i,j]
            if type(poly) is not str:#poly is np.nan:
                continue
            else:
                hqi = df_match.iloc[i,14+j]
                poly_idx = df_label_table.query('polymer == @poly').iloc[0,0]
                ref_id = df_match.iloc[i,28+poly_idx]
                _ = plt.plot(wn, 
                             df_spectra.query('id == @id_').iloc[:,:-6].T, 
                             label='unkown')

                _ = plt.plot(wn, 
                             df_spectra_ref.query('id == @ref_id').iloc[:,:-4].T, 
                             label=str(1+j)+' - '+poly+' - '+str(round(hqi, 4)))
                plt.legend()
                plt.title(id_)
                #plt.show()
                # Save this figure to the PDF
                pdf.savefig()
                plt.close()  # Close figure to free memory
print("PDF saved as "+ file_name_ids[:-4]+".pdf")
```

# Raman

```python
directory_name = 'results/20251026/'
file_name_ids = 'Active_PTIR_Only_Rest_Multiple.csv'
file_name_cp_results = 'CP_results_active_unpaired_raman_LikelyMPs.csv'
file_name_spectra_data = 'active_data_raman_specy.csv'
file_name_reference = 'slopp_data_raman_final_specy.csv'
```

```python
df_ids_plot = pd.read_csv(directory_name+file_name_ids)# IDs you want to plot
ids_plot = df_ids_plot['id'].to_numpy()
n = ids_plot.shape[0]

df_match = pd.read_csv(directory_name+file_name_cp_results) # CP results
del df_match['Unnamed: 0']
df_match = df_match.query('id in @ids_plot').reset_index(drop=True)

df_spectra = pd.read_csv(directory_name+file_name_spectra_data) # Raman data (active or passive)
del df_spectra['Unnamed: 0']

df_label_table = pd.read_csv(directory_name+'label_table_ir.csv')
del df_label_table['Unnamed: 0']

df_spectra_ref = pd.read_csv(directory_name+file_name_reference)
del df_spectra_ref['Unnamed: 0']

df_spectra_ref_mean = df_spectra_ref.drop(['map_x','map_y', 'id'], axis=1).groupby('polymer').mean()
df_spectra_ref_std = df_spectra_ref.drop(['map_x','map_y', 'id'], axis=1).groupby('polymer').std()

wn = np.float64(df_spectra_ref.columns[:-4])
```

```python
init_top = 1 # should be at least 1 and less that last_top
last_top = 14 # should be at most 14

with PdfPages(file_name_ids[:-4]+".pdf") as pdf:
    for i in range(n):
        id_ = df_match['id'][i]
        for j in range(init_top-1, last_top):
            poly = df_match.iloc[i,j]
            if type(poly) is not str:
                continue
            else:
                hqi = df_match.iloc[i,14+j]
                _ = plt.plot(wn, 
                             df_spectra.query('id == @id_').iloc[:,:-5].T, 
                             label='unkown')

                _ = plt.plot(wn, 
                             df_spectra_ref_mean.loc[poly,:].T, 
                             label=str(1+j)+' - '+poly+' - '+str(round(hqi, 4)),
                             color='orange')
                _ = plt.fill_between(wn, 
                                     df_spectra_ref_mean.loc[poly,:].T-df_spectra_ref_std.loc[poly,:].T,
                                     df_spectra_ref_mean.loc[poly,:].T+df_spectra_ref_std.loc[poly,:].T,
                                     alpha=.1,
                                     color='orange')

                plt.legend()
                plt.title(id_)
                #plt.show()
                # Save this figure to the PDF
                pdf.savefig()
                plt.close()  # Close figure to free memory
print("PDF saved as "+ file_name_ids[:-4]+".pdf")
```
