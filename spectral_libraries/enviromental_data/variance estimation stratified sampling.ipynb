```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

L = 10
N_h = 4
N = L * N_h
#n_h
plt.plot(range(1, N_h+1),
    [L*((N_h/N)**2)*(1/n_h)*(1-n_h/N_h) for n_h in range(1, N_h+1)], label='4 schools')


L = 10
N_h = 3
N = L * N_h

plt.plot(range(1, N_h+1),
    [L*((N_h/N)**2)*(1/n_h)*(1-n_h/N_h) for n_h in range(1, N_h+1)], label='3 schools')

plt.ylabel('Expected Variance')
plt.xlabel('# of schools analysed in each month')
plt.legend()
```
